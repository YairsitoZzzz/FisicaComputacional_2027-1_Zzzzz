#### 1.  Cordenadas polares y cartesianas: Supongamos que se nos da (es decri, el usuario introduce)  la posicion de un punto en un espacio bidimensional:


**a)**  Si el punto especifica en coordenadas polares $r, \theta$. Escribe un programa (en 'python' sin usar la biblioteca 'numpy') que calcule e imprima la conversión a coordenadas cartesianas $x, y$.

```python
import math as mt
r=float(input("Valor para R:(grados) "))
theta= float(input("Valor para theta:(grados) "))

x= r* mt.cos((theta*(mt.pi/180)))
y= r * mt.sin(theta*(mt.pi/180))

print("Valor en x:", x)
print("Valor en y:", y)
```

**b)** Si el punto se especifica en coordenadas cartesianas $x,y$. Escribe un programa que calcule e imprima la conversion en coordenadas polares $ r ,\theta $

```python
import math as mt
x_0= float(input("Valor de x"))
y_0= float(input("Valor de y"))

R_0= mt.sqrt(x_0**2 + y_0**2)
theta_0= mt.atan2(y_0,x_0)*180/mt.pi
print("Valor para R=", R_0)
print("Valor para Theta=", theta_0)
```
